## Ontwerp
Een waterbad met bekende massa aan water wordt verwarmd met een elektrisch verwarmingselement dat een bekende hoeveelheid energie levert.

## Data analyse
Geef kort de data-analysemethode weer.

als de waardes van de temperatuur en de tijd bekend zijn kan d.m.v de formule $Q=c*m*\mathrm{\Delta}T$ de soortelijke warmte bepaald worden als $
c=Q/m*\mathrm{\Delta}T$

# Resultaten


````python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.optimize import curve_fit

# data 
T = np.array
T = np.array([17.4,17.4,17.4,17.4,17.5,17.5,17.5,17.5,17.6,17.6,17.6,17.7,17.7,
              17.8,17.8,17.9,17.9,17.9,18.0,18.1,18.2,18.2,18.3,18.3,18.4,18.4,
              18.5,18.6,18.7,18.8,18.9,19.0,19.1,19.3,19.5,19.7,19.8])

time = np.arange(0,181,5)

m = 0.5    #kg
U = 10     #V
I = 1      #A
P = U*I    #W

d_T = T-T[0]

# fitten en onzekerheid 
def fit(t, c):
    return (P * t) / (m * c)

val, cov = curve_fit(fit, time, d_T)
c_fit = val
u_c_fit = np.sqrt(cov)
print("c =", np.round(c_fit,0),"±", np.round(u_c_fit,0), "J/(kg K)")


#plotten van data met functiefit 
t_fit = np.linspace(0, time[-1], 500)
dT_fit = fit(t_fit, c_fit)
plt.figure
plt.scatter(time, d_T, label="Meetdata", color="blue")
plt.plot(t_fit, dT_fit, label="Curve fit", color="red")

plt.xlabel("Tijd (s)")
plt.ylabel("ΔT (°C)")
plt.title("Temperatuurstijging")
plt.legend()

plt.show()
````
